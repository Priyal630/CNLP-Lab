Experiment 4.1: Term-Frequency Analysis and NER Using a Toolkit

```python
from collections import Counter
import csv
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import spacy

# Download necessary NLTK data
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab') # Added to resolve LookupError for punkt_tab

# Read input text file
with open('4.1_4.2_input.txt', 'r', encoding='utf-8') as f:
  text = f.read()

# Tokenization, lowercase conversion, and stop word removal using NLTK
tokens = [word.lower() for word in word_tokenize(text) if word.isalnum()]
stop_words = set(stopwords.words('english'))
filtered_tokens = [w for w in tokens if w not in stop_words]

# Calculate term frequency
word_counts = Counter(filtered_tokens)

# Save term frequencies to CSV format (Term, Frequency)
with open('term_frequency_toolkit.csv', 'w', newline='', encoding='utf-8') as f:
  writer = csv.writer(f)
  writer.writerow(['Term', 'Frequency'])
  for term, freq in word_counts.most_common():
    writer.writerow([term, freq])

# Display top 10 most frequent terms
print('Top 10 Terms (Toolkit):')
for term, freq in word_counts.most_common(10):
  print(f'{term}: {freq}')

# Named Entity Recognition (NER) using spaCy
nlp = spacy.load('en_core_web_sm')
doc = nlp(text)
print('\nNamed Entities (Sample):')
for ent in doc.ents[:10]:
  print(f'{ent.text} -> {ent.label_}')
```

Experiment 4.2: Term-Frequency Analysis Without a Toolkit

```python
from collections import Counter
import csv
import re

# Read input text file[cite: 3]
with open('4.1_4.2_input.txt', 'r', encoding='utf-8') as f:
  text = f.read()

# Convert text to lowercase and extract words using regular expressions (pure Python)
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

# Calculate word frequencies
word_counts = Counter(words)

# Save output in CSV format (Term, Frequency)
with open('term_frequency_without_toolkit.csv', 'w', newline='', encoding='utf-8') as f:
  writer = csv.writer(f)
  writer.writerow(['Term', 'Frequency'])
  for term, freq in word_counts.most_common():
    writer.writerow([term, freq])

# Display top 10 most frequent terms
print('Top 10 Terms (Without Toolkit):')
for term, freq in word_counts.most_common(10):
  print(f'{term}: {freq}')
```

Experiment 4.3: Manual TF-IDF Calculation Across Multiple Documents

```python
import math
import re

# Given input documents[cite: 2]
docs = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence.",
]

# Step 1-3: Convert to lowercase, remove punctuation, and tokenize without toolkits[cite: 2]
tokenized_docs = []
for doc in docs:
  tokens = re.findall(r'\b[a-zA-Z]+\b', doc.lower())
  tokenized_docs.append(tokens)

# Step 4: Calculate TF (Term Frequency) for every term[cite: 2]
tf_list = []
for tokens in tokenized_docs:
  tf_doc = {}
  total_terms = len(tokens)
  counts = Counter(tokens)
  for term, count in counts.items():
    tf_doc[term] = count / total_terms
  tf_list.append(tf_doc)

# Step 5: Calculate DF (Document Frequency) for every term[cite: 2]
df_dict = {}
total_docs = len(docs)
all_terms = set(term for tokens in tokenized_docs for term in tokens)
for term in all_terms:
  df = sum(1 for tokens in tokenized_docs if term in tokens)
  df_dict[term] = df

# Step 6: Calculate IDF using the logarithmic formula[cite: 2]: log10(Total Docs / DF)
idf_dict = {}
for term, df in df_dict.items():
  idf_dict[term] = math.log10(total_docs / df)

# Step 7: Calculate TF-IDF for every term in every document[cite: 2]
tf_idf_list = []
for i, tf_doc in enumerate(tf_list):
  tf_idf_doc = {}
  for term, tf_val in tf_doc.items():
    tf_idf_doc[term] = tf_val * idf_dict[term]
  tf_idf_list.append(tf_idf_doc)

# Step 8-10: Display Term, TF, DF, IDF, TF-IDF and top 10 terms per document[cite: 2]
for i, tf_idf_doc in enumerate(tf_idf_list):
  print(f'\n--- Document {i+1} Top Terms ---')
  sorted_terms = sorted(tf_idf_doc.items(), key=lambda x: x[1], reverse=True)
  for term, score in sorted_terms[:10]:
    print(
        f'Term: {term:<12} | TF: {tf_list[i][term]:.3f} | DF: {df_dict[term]} |'
        f' IDF: {idf_dict[term]:.3f} | TF-IDF: {score:.3f}'
    )
```
