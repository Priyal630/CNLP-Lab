5.1. Subword-Level Tokenization Using a BPE-Based Tokenizer

a. Using a Pretrained Model (transformers)

```python
from transformers import GPT2Tokenizer

# Initialize the pretrained BPE tokenizer (e.g., GPT-2)
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

# Input text sample from the corpus
text = "Natural language processing is a branch of artificial intelligence."

# 1. Tokenize into subword units
tokens = tokenizer.tokenize(text)

# 2. Display generated tokens and corresponding token IDs
token_ids = tokenizer.encode(text)

print("Tokens:", tokens)
print("Token IDs:", token_ids)
```

b. Without Using a Pretrained Model (tokenizers library from HuggingFace)

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

# Initialize a blank BPE model
tokenizer = Tokenizer(BPE(unk_token="<unk>"))
tokenizer.pre_tokenizer = Whitespace()

# Train the tokenizer on the provided input text file
trainer = BpeTrainer(vocab_size=500, special_tokens=["<unk>"])
files = ["input_sub_word_data.txt"]
tokenizer.train(files, trainer)

# 1. Tokenize text into subword units
text = "Natural language processing is a branch of artificial intelligence."
output = tokenizer.encode(text)

# 2. Display generated tokens and corresponding token IDs
print("Tokens:", output.tokens)
print("Token IDs:", output.ids)
```

5.2. Subword-Level Tokenization Using a SentencePiece Tokenizer

a. Using a Pretrained Model (transformers / T5)

```python
from transformers import T5Tokenizer

# Initialize pretrained SentencePiece tokenizer
tokenizer = T5Tokenizer.from_pretrained("t5-small")

text = "Natural language processing is a branch of artificial intelligence."

# 1. Tokenize into subword units
tokens = tokenizer.tokenize(text)

# 2. Display generated tokens and corresponding token IDs
token_ids = tokenizer.encode(text)

print("Tokens:", tokens)
print("Token IDs:", token_ids)
```

b. Without Using a Pretrained Model (sentencepiece)

```python
import sentencepiece as spm

# Train SentencePiece model from the input file
spm.SentencePieceTrainer.train(
    input='input_sub_word_data.txt',
    model_prefix='m',
    vocab_size=500,
    model_type='bpe'
)

# Load the trained model
sp = spm.SentencePieceProcessor()
sp.load('m.model')

text = "Natural language processing is a branch of artificial intelligence."

# 1. Tokenize into subword units
tokens = sp.encode_as_pieces(text)

# 2. Display generated tokens and corresponding token IDs
token_ids = sp.encode_as_ids(text)

print("Tokens:", tokens)
print("Token IDs:", token_ids)
```

5.3. Part-of-Speech (POS) Tagging on a Given Text
Target Sentence: "The young student is reading an interesting book in the library."

a. Using spaCy

```python
import spacy

# Load English language model
nlp = spacy.load("en_core_web_sm")
text = "The young student is reading an interesting book in the library."
doc = nlp(text)

print(f"{'TOKEN':<15} | {'POS TAG':<10} | {'DESCRIPTION'}")
print("-" * 50)
for token in doc:
    print(f"{token.text:<15} | {token.pos_:<10} | {spacy.explain(token.pos_)}")
```

b. Using NLTK

```python
import nltk
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('tagsets')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('tagsets_json')

text = "The young student is reading an interesting book in the library."
tokens = nltk.word_tokenize(text)
pos_tags = nltk.pos_tag(tokens)

print(f"{'TOKEN':<15} | {'POS TAG':<10} | {'DESCRIPTION'}")
print("-" * 50)
for token, tag in pos_tags:
    # Get human-readable description of the tag using NLTK help
    description = nltk.help.upenn_tagset(tag)
```

5.4. POS Tagging with Frequencies Using spaCy
Input Data Source: input_sub_word_data.txt

```python
import spacy
from collections import Counter

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Read input data file
with open("input_sub_word_data.txt", "r", encoding="utf-8") as f:
    text = f.read()

doc = nlp(text)

# Extract unique tokens, their POS tags, descriptions, and compute frequencies
token_data = []
for token in doc:
    if not token.is_space and not token.is_punct:
        token_data.append((token.text, token.pos_, spacy.explain(token.pos_)))

# Calculate frequencies of unique (token, pos, description) tuples
frequency_counts = Counter(token_data)

print(f"{'TOKEN':<20} | {'POS':<6} | {'FREQUENCY':<10} | {'DESCRIPTION'}")
print("-" * 65)
for (token, pos, desc), freq in frequency_counts.most_common():
    print(f"{token:<20} | {pos:<6} | {freq:<10} | {desc}")
```
