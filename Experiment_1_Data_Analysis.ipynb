# **Experiment-1 Data Analysis**

## Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/MyDrive/CNLP-Lab/Experiment-1/employee_information_100.csv')

avg_salary= df.groupby("Department")["Salary"].mean()

print(avg_salary)


avg_salary.plot(kind="bar",color="green")
plt.title("Average Salary of each Departments")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.show()
```

## Experiment 1.2: Display the number of employees working in each department.



```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/MyDrive/CNLP-Lab/Experiment-1/employee_information_100.csv')
dept_count = df["Department"].value_counts()

dept_count.plot(kind="bar", color="orange")
plt.title("Employees in Each Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.show()
```

## Experiment 1.3:Display the percentage of male and female employees using a pie chart


```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/MyDrive/CNLP-Lab/Experiment-1/employee_information_100.csv')
gender_count = df["Gender"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(gender_count,
        labels=gender_count.index,
        autopct="%1.1f%%",
        startangle=90)
plt.title("Gender Distribution")
plt.show()
```

## Experiment 1.4: Draw a histogram showing salary distribution.


```python
plt.figure(figsize=(8,5))
plt.hist(df["Salary"], bins=10, edgecolor="black", label="Salary")
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

## Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Experience_Years"], df["Salary"])
plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.show()
```

## Experiment 1.6: Display the top 10 employees with the highest salary.


```python
top10 = df.nlargest(10, "Salary")

print("\nTop 10 Highest Paid Employees")
top10[["Employee_ID", "Name", "Department", "Salary"]]
```

## Experiment 1.7: Find the highest salary in every department.


```python
highest_salary = df.groupby("Department")["Salary"].max()

print("\nHighest Salary in Each Department")
highest_salary
```

## Experiment 1.8: Display employees whose salary is greater than the overall average salary.

```python
overall_avg = df["Salary"].mean()

above_avg = df[df["Salary"] > overall_avg]

print("\nOverall Average Salary:", overall_avg)
print("\nEmployees Earning Above Average Salary")
above_avg[["Employee_ID", "Name", "Department", "Salary"]]
```


## Experiment 1.9: Calculate the average years of experience in each department.


```python
avg_exp = df.groupby("Department")["Experience_Years"].mean()

print("\nAverage Experience in Each Department")
avg_exp
```
